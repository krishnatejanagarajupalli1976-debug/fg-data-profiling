```python
array_1d = np.array([1, 2, 3, 4, 5])

array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D array:")
print(array_1d)
print("Shape:", array_1d.shape)

print("\n2D array:")
print(array_2d)
print("Shape:", array_2d.shape)

print("\n3D array:")
print(array_3d)
print("Shape:", array_3d.shape)
```

```python
array = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

addition = array + 5

print("Original array:")
print(array)

print("\nAfter broadcasting:")
print(addition)
```

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

addition = a + b
multiplication = a * b
square = a ** 2

print("Addition:", addition)
print("Multiplication:", multiplication)
print("Square:", square)
```

```python
matrix_a = np.array([
    [1, 2],
    [3, 4]
])

matrix_b = np.array([
    [5, 6],
    [7, 8]
])

result = matrix_a @ matrix_b

print("Matrix A:")
print(matrix_a)

print("\nMatrix B:")
print(matrix_b)

print("\nMatrix multiplication result:")
print(result)
```

```python
import numpy as np
import pandas as pd

file_path = "docs/getting-started/data_quality_alerts.csv"

df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())

print("\nColumn names:")
print(df.columns.tolist())

# Select numeric columns
numeric_df = df.select_dtypes(include=np.number)

print("\nNumeric columns:")
print(numeric_df.columns.tolist())

# Mean using NumPy
mean_values = np.mean(numeric_df, axis=0)

# Standard deviation using NumPy
std_values = np.std(numeric_df, axis=0)

# Correlation using NumPy
correlation_matrix = np.corrcoef(numeric_df.T)

print("\nMean:")
print(mean_values)

print("\nStandard Deviation:")
print(std_values)

print("\nCorrelation Matrix:")
print(correlation_matrix)
```

```python
import numpy as np
import pandas as pd

# Real Iris CSV dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df_iris = pd.read_csv(url)

print("Dataset shape:", df_iris.shape)

print("\nFirst 5 rows:")
print(df_iris.head())

# Select numeric columns
numeric_data = df_iris.select_dtypes(include=np.number)

print("\nNumeric columns:")
print(numeric_data.columns.tolist())

# NumPy calculations
mean_values = np.mean(numeric_data, axis=0)
std_values = np.std(numeric_data, axis=0)
correlation_matrix = np.corrcoef(numeric_data.T)

print("\nMean:")
for column, value in zip(numeric_data.columns, mean_values):
    print(f"{column}: {value:.4f}")

print("\nStandard Deviation:")
for column, value in zip(numeric_data.columns, std_values):
    print(f"{column}: {value:.4f}")

print("\nCorrelation Matrix:")
print(pd.DataFrame(
    correlation_matrix,
    index=numeric_data.columns,
    columns=numeric_data.columns
))
```

```python
import numpy as np
import pandas as pd

# Load the real Iris dataset
iris_df = pd.read_csv("iris.txt")

# Select only numeric columns
numeric_data = iris_df.select_dtypes(include=np.number)

print("Numeric columns:")
print(numeric_data.columns.tolist())

# Calculate mean using NumPy
mean_values = np.mean(numeric_data, axis=0)

# Calculate standard deviation using NumPy
std_values = np.std(numeric_data, axis=0)

# Calculate correlation using NumPy
correlation_matrix = np.corrcoef(numeric_data.T)

# Display mean
print("\nMean:")
for column, value in zip(numeric_data.columns, mean_values):
    print(f"{column}: {value:.4f}")

# Display standard deviation
print("\nStandard Deviation:")
for column, value in zip(numeric_data.columns, std_values):
    print(f"{column}: {value:.4f}")

# Display correlation matrix
print("\nCorrelation Matrix:")
correlation_df = pd.DataFrame(
    correlation_matrix,
    index=numeric_data.columns,
    columns=numeric_data.columns
)
print(correlation_df.round(4))
```
